```python
from tkinter import Tk, filedialog
from pathlib import Path

# Crear intro.ipynb en docs con una celda de markdown
import nbformat as nbf

# Pedir nombre del proyecto
nombre_proyecto = input("Introduce el nombre del proyecto: ")

# Abrir diálogo para seleccionar carpeta base
Tk().withdraw()
carpeta_base = filedialog.askdirectory(title="Selecciona la carpeta donde crear el proyecto")

# Crear ruta del proyecto
ruta_proyecto = Path(carpeta_base) / nombre_proyecto
ruta_proyecto.mkdir(parents=True, exist_ok=True)

# Crear subdirectorios
(ruta_proyecto / "docs").mkdir(exist_ok=True)
(ruta_proyecto / "python").mkdir(exist_ok=True)

# Crear archivos
with open(ruta_proyecto / "LEEME.txt", "w", encoding="utf-8") as f:
    f.write(nombre_proyecto)

intro_nb = nbf.v4.new_notebook()
intro_nb.cells.append(nbf.v4.new_markdown_cell(f"# {nombre_proyecto}"))

with open(ruta_proyecto / "docs" / "intro.ipynb", "w", encoding="utf-8") as f:
    nbf.write(intro_nb, f)

# Crear app.py vacío en python
open(ruta_proyecto / "python" / "app.py", "w", encoding="utf-8").close()

print(f"Estructura del proyecto '{nombre_proyecto}' creada correctamente en {ruta_proyecto}")
```
