```
import pandas as pd
from tkinter import Tk, filedialog, simpledialog
import smtplib
from email.message import EmailMessage

# Abrir cuadro de diálogo para seleccionar archivo CSV
Tk().withdraw()
ruta_csv = filedialog.askopenfilename(filetypes=[("CSV files", "*.csv")], title="Selecciona el archivo CSV de clientes")

# Cargar datos de clientes
clientes = pd.read_csv(ruta_csv)
clientes.head()
```

```
# Pedir encabezado y cuerpo del mensaje
Tk().withdraw()
asunto = simpledialog.askstring("Encabezado", "Introduce el encabezado del mensaje (usa {nombre} para personalizar):")
cuerpo = simpledialog.askstring("Cuerpo", "Introduce el cuerpo del mensaje (usa {nombre} para personalizar):")
```

```
# Configuración del servidor SMTP (modifica con tus datos reales)
EMAIL_REMITENTE = "tucorreo@ejemplo.com"
PASSWORD = "tu_contraseña"
SMTP_SERVER = "smtp.ejemplo.com"
SMTP_PORT = 587

# Crear conexión
with smtplib.SMTP(SMTP_SERVER, SMTP_PORT) as smtp:
    smtp.starttls()
    smtp.login(EMAIL_REMITENTE, PASSWORD)

    for _, cliente in clientes.iterrows():
        msg = EmailMessage()
        msg['From'] = EMAIL_REMITENTE
        msg['To'] = cliente['email']
        msg['Subject'] = asunto.format(nombre=cliente['nombre'])
        msg.set_content(cuerpo.format(nombre=cliente['nombre']))

        smtp.send_message(msg)
        print(f"Correo enviado a {cliente['nombre']} ({cliente['email']})")
```
