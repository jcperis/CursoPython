# 📘 Tutorial de NumPy en Python: Características Más Importantes

NumPy (**Numerical Python**) es una biblioteca fundamental para la computación científica en Python. Proporciona:
- Un potente objeto array N-dimensional (`ndarray`)
- Herramientas para álgebra lineal, estadísticas, generación de números aleatorios, etc.

## 📌 2. Crear arrays con NumPy

```python
import numpy as np

# Array desde lista
a = np.array([1, 2, 3])
print(a)

# Array bidimensional
b = np.array([[1, 2], [3, 4]])
print(b)
```

```python
print(np.zeros((2, 3)))
print(np.ones((3,)))
print(np.full((2, 2), 7))
print(np.eye(3))
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
```

## 🔄 3. Operaciones básicas

```python
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print(x + y + x)
print(x * y)
print(x * np.transpose(x))
print(np.sin(x))
```

## 🔍 4. Indexado y slicing

```python
a = np.array([[10, 20, 30], [40, 50, 60]])

print(a[0, 1])
print(a[:, 1])
print(a[1, :])
print(a[0:2, 1:3])
```

## 🔁 5. Forma (shape), tamaño (size), tipo (dtype)

```python
a = np.array([[1, 2], [3, 4]])

print(a.shape)
print(a.size)
print(a.dtype)
```

## 🔄 6. Cambiar forma (reshape)

```python
a = np.arange(6)
b = a.reshape((2, 3))
print(b)
```

## 🔁 7. Concatenación y división

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])

print(np.vstack((a, b)))
print(np.hstack((a, b.T)))

x = np.array([1, 2, 3, 4, 5, 6])
print(np.split(x, 3))
```

## 🎲 8. Números aleatorios

```python
np.random.seed(0)
print(np.random.rand(2, 2))
print(np.random.randn(2, 2))
print(np.random.randint(1, 10))
```

## 📐 9. Álgebra lineal

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[2, 0], [1, 2]])

print(np.dot(a, b))
print(a @ b)
print(a.T)
print(np.linalg.inv(a))
print(np.linalg.det(a))
eigenvalues, eigenvectors = np.linalg.eig(a)
print(eigenvalues)
print(eigenvectors)
```

## 📊 10. Estadísticas

```python
a = np.array([1, 2, 3, 4, 5])

print(np.mean(a))
print(np.median(a))
print(np.std(a))
print(np.var(a))
print(np.min(a))
print(np.max(a))
```

## 🎯 11. Comparaciones y máscaras booleanas

```python
a = np.array([1, 2, 3, 4, 5])

mask = a > 2
print(a[mask])
print(a[a % 2 == 0])
```

## 🧰 12. Funciones útiles

```python
print(np.unique([1, 1, 2, 3]))
print(np.sort([4, 1, 3, 2]))
print(np.argsort([4, 1, 3, 2]))
print(np.argmax([1, 3, 7, 2]))
print(np.cumsum([1, 2, 3]))
print(np.cumprod([1, 2, 3, 4]))
```

## 🧪 13. Broadcasting

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([10, 20, 30])

print(a + b)
```

## 🧼 14. Evitar errores comunes

```python
a = np.array([1, 2, 3])
b = a
b[0] = 10
print(a)

b = a.copy()
b[0] = 99
print(a)
```
